## Marco Conceptual y Preparación del Entorno

### La Ruta Relativa: `datos/registro.txt`
Una ruta relativa se define a partir del directorio donde se ejecuta el programa, lo que hace al código portable entre distintos sistemas operativos (Linux, macOS, Windows).

```python
from pathlib import Path
from typing import Final

CARPETA: Final[Path] = Path("datos")
ARCHIVO: Final[Path] = CARPETA / "registro.txt"

print(f"Carpeta definida: {CARPETA}")
print(f"Ruta del archivo: {ARCHIVO}")
```

## Resolución Paso a Paso de los 11 Ejercicios

### Ejercicio 1. Crear la carpeta y la ruta

Se importa `Path` desde la biblioteca estándar `pathlib`. El operador `/` combina el directorio `datos` con el nombre de archivo `registro.txt`, generando un objeto de ruta multiplataforma.

```python
from pathlib import Path

CARPETA = Path("datos")
ARCHIVO = CARPETA / "registro.txt"

print(ARCHIVO)
```

### Ejercicio 2. Crear la carpeta si no existe

El método `CARPETA.mkdir(exist_ok=True)` crea el directorio físico en el disco. El parámetro `exist_ok=True` es defensivo: si la carpeta ya existe, no arroja ninguna excepción y continúa la ejecución.

```python
CARPETA.mkdir(exist_ok=True)
print(f"¿Existe la carpeta '{CARPETA}' en disco? {CARPETA.is_dir()}")
```

### Ejercicio 3. Crear el archivo TXT sin truncar

Se utiliza el modo de apertura `'a'` (*Append*). Si el archivo no existe, Python lo crea de inmediato. Si ya existe, no modifica su contenido. La instrucción `pass` expresa formalmente que en este paso solo preparamos el archivo sin escribir aún datos.

```python
with open(ARCHIVO, "a", encoding="utf-8") as archivo:
    pass

print(f"¿Existe el archivo '{ARCHIVO}'? {ARCHIVO.is_file()}")
```

### Ejercicio 4. Escribir una primera línea

Para almacenar texto se utiliza `archivo.write()`. Se concatena o formatea explícitamente el salto de línea `\n` mediante una f-string (`f"{texto}\n"`). Sin el salto `\n`, dos escrituras sucesivas quedarían unidas en la misma línea física.

```python
texto = "Primera prueba de archivo"

with open(ARCHIVO, "a", encoding="utf-8") as archivo:
    archivo.write(f"{texto}\n")

print(f"Línea escrita: '{texto}'")
```

### Ejercicio 5. Comprobar que agregar no significa reemplazar

Al escribir una segunda frase en modo `'a'`, Python posiciona el puntero al final del archivo. La primera frase se conserva intacta y la nueva línea se agrega inmediatamente abajo. Si hubiésemos usado modo `'w'`, la primera frase se habría borrado.

```python
segunda_frase = "Segunda prueba de persistencia"

with open(ARCHIVO, "a", encoding="utf-8") as archivo:
    archivo.write(f"{segunda_frase}\n")

print(f"Segunda línea agregada: '{segunda_frase}'")
```

### Ejercicio 6. Solicitar un dato con input() y registrar un saludo

La función `input()` captura una entrada del usuario como cadena (`str`). Se aplica `.strip()` para limpiar espacios accidentales y se escribe en el archivo utilizando una f-string formateada (`f"Hola, {nombre}\n"`).

```python
nombre_prueba = "Ana"  # Simulación de entrada de usuario
nombre_limpio = nombre_prueba.strip()

with open(ARCHIVO, "a", encoding="utf-8") as archivo:
    archivo.write(f"Hola, {nombre_limpio}\n")

print(f"Saludo guardado para: {nombre_limpio}")
```

### Ejercicio 7. Registrar un producto

En esta etapa introductoria, se guarda únicamente el nombre del producto en una línea simple, sentando las bases antes de incorporar campos múltiples o delimitadores.

```python
producto_prueba = "Silla"

with open(ARCHIVO, "a", encoding="utf-8") as archivo:
    archivo.write(f"{producto_prueba.strip()}\n")

print(f"Producto registrado: '{producto_prueba}'")
```

### Ejercicio 8. Agregar varios datos con un ciclo for

Un bucle `for` permite repetir la operación de captura y escritura para múltiples elementos. Cada iteración anexa una nueva línea al archivo sin borrar los registros acumulados.

```python
lote_productos = ["Mesa", "Escritorio", "Lampara"]

with open(ARCHIVO, "a", encoding="utf-8") as archivo:
    for prod in lote_productos:
        archivo.write(f"{prod.strip()}\n")

print(f"Lote de {len(lote_productos)} productos anexado con éxito.")
```

### Ejercicio 9. Leer todas las líneas con modo 'r'

El modo `'r'` (*Read*) abre el archivo en modo de solo lectura. El bucle `for linea in archivo:` recorre el archivo de forma eficiente entregando una línea a la vez. `linea.strip()` remueve el salto de línea `\n` al imprimir, evitando líneas en blanco intermedias.

```python
print("CONTENIDO COMPLETO DE 'datos/registro.txt':")
print("-" * 45)

with open(ARCHIVO, "r", encoding="utf-8") as archivo:
    for linea in archivo:
        print(linea.strip())

print("-" * 45)
```

### Ejercicio 10. Separar escritura y lectura en funciones

Se aplica el principio de **Responsabilidad Única**:
- `preparar_archivo()`: Garantiza la existencia del directorio y archivo.
- `agregar_dato(dato)`: Se encarga exclusivamente de escribir una línea.
- `leer_datos()`: Se encarga exclusivamente de leer y retornar las líneas limpias.

```python
def preparar_archivo_func():
    CARPETA.mkdir(exist_ok=True)
    with open(ARCHIVO, "a", encoding="utf-8"):
        pass

def agregar_dato_func(dato: str):
    preparar_archivo_func()
    with open(ARCHIVO, "a", encoding="utf-8") as f:
        f.write(f"{dato.strip()}\n")

def leer_datos_func():
    if not ARCHIVO.exists():
        return []
    with open(ARCHIVO, "r", encoding="utf-8") as f:
        return [l.strip() for l in f if l.strip()]

agregar_dato_func("Estante")
print(f"Total de líneas leídas mediante función: {len(leer_datos_func())}")
```

### Ejercicio 11. Crear un menú mínimo interactivo

Un bucle continuo `while True` coordina la interacción del usuario:
- Opción 1: Invoca `agregar_dato()`.
- Opción 2: Invoca `leer_datos()` y muestra el contenido.
- Opción 3: Ejecuta `break` para salir limpiamente del programa.

```python
menu_texto = (
    "========================================
"
    "MENU BASICO DE ARCHIVOS TXT
"
    "========================================
"
    "1. Agregar dato
"
    "2. Mostrar datos
"
    "3. Salir
"
    "========================================"
)
print(menu_texto.strip())
print("La lógica se ejecuta dentro de un ciclo while True con despacho if/elif/else.")
```

## Análisis de Errores Frecuentes

1. **Usar `'w'` para agregar registros:** El modo `'w'` trunca a 0 bytes el archivo al abrir, eliminando todo el histórico. Para conservar lo anterior, debe usarse `'a'`.
2. **Olvidar el salto de línea `\n`:** Provoca que los registros se escriban pegados en una sola línea horizontal indistinguible (`SillaMesaEscritorio`).
3. **Intentar leer con `'r'` antes de crear el archivo:** Provoca una excepción `FileNotFoundError` que interrumpe el programa.
4. **No utilizar el Context Manager `with`:** Puede provocar que los datos queden retenidos en el búfer de RAM sin guardarse en disco, o dejar descriptores abiertos en el Kernel.
5. **Crear manualmente el archivo TXT en el explorador:** El programa debe ser autónomo y crear sus carpetas y archivos mediante código Python (`mkdir(exist_ok=True)` y `open(..., "a")`).
6. **Intentar resolver estructuras complejas prematuramente:** Antes de manipular diccionarios o cálculos de ventas, es fundamental dominar el ciclo elemental de creación, anexión y lectura línea por línea.
