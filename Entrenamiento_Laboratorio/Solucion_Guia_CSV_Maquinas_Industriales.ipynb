### Propósito del Ejercicio Integral
Diseñar un sistema en Python para registrar las condiciones de operación de máquinas industriales (Fecha, Máquina, Temperatura, Horas_Trabajo, Estado), aplicando la evaluación automática del estado operacional:
* **Normal:** Temperatura $\le 70$ °C **y** Horas $\le 8$.
* **Advertencia:** Temperatura entre $71$ y $90$ °C, **o** Horas entre $9$ y $12$.
* **Critico:** Temperatura $> 90$ °C **o** Horas $> 12$.

## 1. Declaración de Funciones Modulares de Evaluación y Persistencia

```python
import csv
from pathlib import Path

CARPETA_DATOS = Path("datos")
ARCHIVO_MAQUINAS = CARPETA_DATOS / "registro_maquinas.csv"

def asegurar_archivo():
    CARPETA_DATOS.mkdir(parents=True, exist_ok=True)
    if not ARCHIVO_MAQUINAS.exists():
        with ARCHIVO_MAQUINAS.open("w", newline="", encoding="utf-8") as f:
            escritor = csv.writer(f, delimiter=";")
            escritor.writerow(["Fecha", "Maquina", "Temperatura", "Horas_Trabajo", "Estado"])
        print(" Archivo inicializado con 5 encabezados.")

def evaluar_estado(temperatura: int, horas: int) -> str:
    if temperatura > 90 or horas > 12:
        return "Critico"
    elif (71 <= temperatura <= 90) or (9 <= horas <= 12):
        return "Advertencia"
    else:
        return "Normal"

asegurar_archivo()
```

## 2. Inserción de Revisiones de Prueba

```python
def registrar_revision(fecha: str, maquina: str, temp: int, horas: int):
    asegurar_archivo()
    estado = evaluar_estado(temp, horas)
    with ARCHIVO_MAQUINAS.open("a", newline="", encoding="utf-8") as f:
        escritor = csv.writer(f, delimiter=";")
        escritor.writerow([fecha, maquina, temp, horas, estado])
    print(f"Registrado: {maquina} | Temp: {temp}°C | Horas: {horas}h | Estado: {estado}")

registrar_revision("2026-08-30", "Torno-01", 65, 7)
registrar_revision("2026-08-30", "Prensa-02", 82, 9)
registrar_revision("2026-08-30", "Cinta-03", 96, 13)
```

## 3. Lectura y Resumen Estadístico Operacional

```python
def mostrar_resumen():
    with ARCHIVO_MAQUINAS.open("r", newline="", encoding="utf-8") as f:
        lector = csv.reader(f, delimiter=";")
        encabezado = next(lector)
        registros = list(lector)
    
    total = len(registros)
    temps = [int(r[2]) for r in registros]
    promedio = sum(temps) / total
    normales = sum(1 for r in registros if r[4] == "Normal")
    advertencias = sum(1 for r in registros if r[4] == "Advertencia")
    criticos = sum(1 for r in registros if r[4] == "Critico")

    print(f"Total Revisiones: {total}")
    print(f"Promedio Temperatura: {promedio:.1f} °C")
    print(f"Estado Normal: {normales} | Advertencia: {advertencias} | Crítico: {criticos}")

mostrar_resumen()
```
