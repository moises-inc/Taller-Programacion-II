### Ejercicio 1. Preparar el proyecto y el espacio de trabajo

Se organiza la estructura del proyecto en Python. El archivo fuente no depende de la creación manual previa del archivo de texto en el explorador de archivos, sino que el propio código gestionará la inicialización de carpetas y archivos.

```python
from pathlib import Path
from typing import Final

print("Espacio de trabajo listo para inicialización programática.")
```

### Ejercicio 2. Crear la ruta del archivo con pathlib.Path

Se define la carpeta contenedora `datos` y la ruta relativa `datos/registro.txt`. Se utiliza el operador de composición `/` de `pathlib.Path` para asegurar compatibilidad en Linux, macOS y Windows.

```python
CARPETA_DATOS: Final[Path] = Path("datos")
ARCHIVO_REGISTRO: Final[Path] = CARPETA_DATOS / "registro.txt"

print(f"Carpeta: {CARPETA_DATOS}")
print(f"Ruta Relativa: {ARCHIVO_REGISTRO}")
```

### Ejercicio 3. Crear el archivo si no existe (Modo 'a')

Se crea el directorio mediante `mkdir(parents=True, exist_ok=True)` y se abre el archivo en modo `'a'` (*Append*). Si no existía, el sistema operativo lo crea vacío; si ya existía, no altera su contenido.

```python
CARPETA_DATOS.mkdir(parents=True, exist_ok=True)

with ARCHIVO_REGISTRO.open("a", encoding="utf-8"):
    pass

print(f"¿Existe el archivo físico? {ARCHIVO_REGISTRO.is_file()}")
```

### Ejercicio 4. Escribir una primera línea con f-string y salto explícito

Se guarda una frase inicial en el archivo de texto. La adición de `\n` garantiza la separación física de registros.

```python
frase_inicial = "Registro inicial de prueba"

with ARCHIVO_REGISTRO.open("a", encoding="utf-8") as f:
    f.write(f"{frase_inicial}\n")

print(f"Línea 1 escrita: '{frase_inicial}'")
```

### Ejercicio 5. Comprobar que la información se conserva entre ejecuciones

Al ejecutar una segunda escritura con una frase distinta en modo `'a'`, la nueva línea se anexa al final sin borrar la anterior.

```python
frase_segunda = "Segunda línea de verificación"

with ARCHIVO_REGISTRO.open("a", encoding="utf-8") as f:
    f.write(f"{frase_segunda}\n")

print(f"Línea 2 escrita: '{frase_segunda}'")
```

### Ejercicio 6. Solicitar información al usuario con input() y registrar saludo

Se captura el nombre de una persona mediante `input().strip()` y se almacena un saludo formateado mediante `f"Hola, {nombre}\n"`.

```python
nombre_usuario = "Carlos"  # Simulación de entrada input()

with ARCHIVO_REGISTRO.open("a", encoding="utf-8") as f:
    f.write(f"Hola, {nombre_usuario}\n")

print(f"Saludo registrado para: '{nombre_usuario}'")
```

### Ejercicio 7. Registrar productos individuales

Se solicita el nombre de un artículo y se anexa en una línea nueva sin delimitadores adicionales.

```python
producto_individual = "Silla Ergonomica"

with ARCHIVO_REGISTRO.open("a", encoding="utf-8") as f:
    f.write(f"{producto_individual.strip()}\n")

print(f"Producto registrado: '{producto_individual}'")
```

### Ejercicio 8. Registrar varios productos en una sola ejecución con ciclo for

Un bucle `for` itera sobre una colección de productos e invoca `.write()` para cada uno de ellos, agregando tres líneas nuevas al archivo.

```python
lote_tres_productos = ["Mesa de Centro", "Escritorio Ejecutivo", "Lampara LED"]

with ARCHIVO_REGISTRO.open("a", encoding="utf-8") as f:
    for prod in lote_tres_productos:
        f.write(f"{prod.strip()}\n")

print(f"Se agregaron {len(lote_tres_productos)} productos en lote.")
```

### Ejercicio 9. Leer el archivo completo en modo 'r'

Se abre el archivo en modo de solo lectura (`'r'`). La iteración `for linea in f:` lee secuencialmente todos los registros acumulados.

```python
print("LECTURA BRUTA DEL ARCHIVO:")
with ARCHIVO_REGISTRO.open("r", encoding="utf-8") as f:
    for linea in f:
        print(repr(linea))
```

### Ejercicio 10. Mostrar las líneas limpias sin saltos adicionales (.strip())

Al aplicar `linea.strip()`, se remueve el `\n` residual al final de la cadena, evitando que la función `print()` inserte una línea en blanco intermedia no deseada.

```python
print("REPORTE LIMPIO DE REGISTROS:")
print("-" * 40)
with ARCHIVO_REGISTRO.open("r", encoding="utf-8") as f:
    for linea in f:
        linea_limpia = linea.strip()
        if linea_limpia:
            print(linea_limpia)
print("-" * 40)
```

### Ejercicio 11. Contar la cantidad total de registros almacenados

Se implementa el patrón acumulador para calcular el número total de líneas no vacías almacenadas en el archivo TXT.

```python
conteo_total = 0

with ARCHIVO_REGISTRO.open("r", encoding="utf-8") as f:
    for linea in f:
        if linea.strip():
            conteo_total += 1

print(f"Total de registros contabilizados en disco: {conteo_total}")
```

### Ejercicio 12. Numerar los registros correlativamente con enumerate()

La función `enumerate(archivo, start=1)` genera automáticamente un índice numérico comenzando desde 1 para cada registro.

```python
print("REGISTROS NUMERADOS CORRELATIVAMENTE:")
print("=" * 45)

with ARCHIVO_REGISTRO.open("r", encoding="utf-8") as f:
    for numero, linea in enumerate(f, start=1):
        print(f"[{numero:02d}] {linea.strip()}")

print("=" * 45)
```

### Ejercicio 13. Crear una función dedicada para agregar datos

Se encapsula la lógica de escritura en la función `agregar_dato(texto: str) -> None`, garantizando que la preparación del archivo y el modo `'a'` estén centralizados.

```python
def agregar_dato(texto: str) -> None:
    # Anexa una línea de texto al archivo persistente.
    CARPETA_DATOS.mkdir(parents=True, exist_ok=True)
    with ARCHIVO_REGISTRO.open("a", encoding="utf-8") as f:
        f.write(f"{texto.strip()}\n")

agregar_dato("Estante Modular")
print("Función agregar_dato() ejecutada exitosamente.")
```

### Ejercicio 14. Crear una función dedicada para leer y recuperar datos

Se encapsula la lectura en `leer_datos() -> list[str]`, retornando una lista con los registros procesados para su uso en cualquier parte del sistema.

```python
def leer_datos() -> list[str]:
    # Retorna la lista de líneas limpias almacenadas en el archivo.
    if not ARCHIVO_REGISTRO.exists():
        return []
    with ARCHIVO_REGISTRO.open("r", encoding="utf-8") as f:
        return [l.strip() for l in f if l.strip()]

datos_recuperados = leer_datos()
print(f"Registros recuperados por la función: {len(datos_recuperados)}")
print(f"Último registro anexado: '{datos_recuperados[-1]}'")
```

### Ejercicio 15. Integrar las operaciones en un menú interactivo continuo

Se construye la interfaz de consola mediante un ciclo `while True`, `input()` y una estructura condicional `if/elif/else`, permitiendo repetir acciones de forma fluida.

```python
def mostrar_menu_interactivo():
    print("\n" + "=" * 40)
    print("SISTEMA DE CONTROL DE ARCHIVOS TXT")
    print("=" * 40)
    print("  1. Agregar dato")
    print("  2. Mostrar datos guardados")
    print("  3. Mostrar registros numerados y total")
    print("  4. Salir")
    print("=" * 40)

mostrar_menu_interactivo()
print("El menú delega cada acción a sus funciones especializadas y finaliza con la opción 4.")
```
